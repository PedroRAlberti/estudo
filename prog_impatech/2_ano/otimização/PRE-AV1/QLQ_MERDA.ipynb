```python
from pulp import *
import matplotlib.pyplot as plt
import numpy as np
```

```python
lp = LpProblem("milk-problem",LpMaximize) #problema 3.b

F = LpVariable(name="A",  lowBound=0,cat="Continuous")
D = LpVariable(name="D", lowBound=0, cat="Continuous")
B = LpVariable(name="B", lowBound=0,cat="Continuous")

lp += 0.043*A + 0.022*D - 0.0275*B

lp += A + D - B <= 10
lp += B <= 1
lp += A <= 3
lp += D >= 4
lp += 3*A - 2*D <= 0
lp += 2*A - D <= 0

status = lp.solve()

for var in lp.variables():
    print(var.name, var.value())
print(lp.objective.value())
```

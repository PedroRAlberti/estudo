## Questão 3

```python
lp = LpProblem("bond-problem",LpMaximize) #problema 3.b

A = LpVariable(name="A",  lowBound=0,cat="Continuous")
D = LpVariable(name="D", lowBound=0, cat="Continuous")
B = LpVariable(name="B", lowBound=0,cat="Continuous")

lp += 0.043*A + 0.022*D - 0.0275*B

lp += A + D - B <= 10
lp += B <= 1
lp += A <= 3
lp += D >= 4
lp += 3*A - 2*D <= 0
lp += 2*A - D <= 0

status = lp.solve()

for var in lp.variables():
    print(var.name, var.value())
print(lp.objective.value())
```

### Problema 5

```python
lp = LpProblem("truck_problem",LpMaximize) #problema 5

A = LpVariable(name="Aztecs", lowBound=0, cat="Integer") #Nomes das variáveis
B = LpVariable(name="Broncos", lowBound=0, cat="Integer")

lp += 590*A + 270*B # Função objetivo

lp += A <= 2500 #condições
lp += B <= 3000
lp += A + B <= 4000
lp += 2*A + B <= 6000

status = lp.solve() 

for var in lp.variables():
    print("deve vender do modelo",var.name, var.value(),"unidades.")
print("lucro total é",lp.objective.value())
```

## Problema 9

```python
lp = LpProblem("camera_problems",LpMaximize) #problema 9

x1 = LpVariable(name="Cub", lowBound=250, cat="Integer") #nomes das variáveis
x2 = LpVariable(name="Quickiematic", lowBound=375, cat="Integer")
x3 = LpVariable(name="VIP", lowBound=150, cat="Integer")

lp += 3*x1 + 9*x2 + 25*x3 # Função objetivo

lp += 0.1*x1 + 0.2*x2 + 0.7*x3 <= 250 #condições
lp += 0.2*x1 + 0.35*x2 + 0.1*x3 <= 350
lp += 0.1*x1 + 0.2*x2 + 0.3*x3 <= 150

status = lp.solve() 

for var in lp.variables():
    print("deve vender do modelo",var.name, var.value(),"unidades.")
print("lucro total é",lp.objective.value())
```

## Problema 11

```python
lp = LpProblem("petroleum-problem",LpMaximize) #problema 11

D_A = LpVariable(name="Deluxe_A", lowBound=0) #nomes das variáveis
D_B = LpVariable(name="Deluxe_B", lowBound=0)
D_C = LpVariable(name="Deluxe_C", lowBound=0)
D   = LpVariable(name="Deluxe",   lowBound=0)

S_A = LpVariable(name="Standard_A", lowBound=0)
S_B = LpVariable(name="Standard_B", lowBound=0)
S_C = LpVariable(name="Standard_C", lowBound=0)
S   = LpVariable(name="Standard",   lowBound=0)

E_A = LpVariable(name="Economy_A", lowBound=0)
E_B = LpVariable(name="Economy_B", lowBound=0)
E_C = LpVariable(name="Economy_C", lowBound=0)
E   = LpVariable(name="Economy",   lowBound=0)

A = D_A + E_A + S_A
B = D_B + E_B + S_B 
C = D_C + E_C + S_C

lp += 7.9*D + 6.9*S + 5*E - (0.60*A + 0.52*B + 0.48*C) # Função objetivo

lp += D_A <= 0.6*D #condições
lp += D_C >= 0.2*D
lp += S_A <= 0.15*S
lp += S_C >= 0.6*S
lp += E_C >= 0.5*E

lp += D_A + E_A + S_A <= 4000
lp += D_B + E_B + S_B <= 5000
lp += D_C + E_C + S_C <= 2500

lp += D == D_A + D_B + D_C
lp += S == S_A + S_B + S_C
lp += E == E_A + E_B + E_C

status = lp.solve()

for var in lp.variables():
    print(var.name, var.value())
print("lucro total é",lp.objective.value())
```

## Problema 12 

```python
lp = LpProblem("aircraft-problem",LpMaximize) #problema 12

A = LpVariable(name="A", lowBound=0) #Nome das variáveis
B = LpVariable(name="B", lowBound=0)
C = LpVariable(name="C", lowBound=0)
E = LpVariable(name="Extern", lowBound=0)

lp += 10*18*350*A + 8*20*300*18*B + 120*18*300*21*C + 0.2*E # função objetivo

lp += A + B + C <= 30 #condições
lp += 80*A + 130*B + 150*C <= 4000
lp += A + 2*B + 2*C <= 50

lp += 10*350*18*A + 20*300*18*B + 18*300*21*C + E == 3500000


status = lp.solve()

for var in lp.variables():
    if var.name != "Extern":
        print("deve comprar do modelo",var.name, var.value(),"unidades.")
    if var.name == "Extern":
        print("deve exportar ",var.value())

print("lucro total é",lp.objective.value())
```
